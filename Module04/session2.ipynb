```python
# Access elements
print("First element:", scores[0])
```

```python
# Iterate
for score in scores:
  print(score)
```

```python
# Sort
scores.sort()
print("Sorted:", scores)
```

```python
scores.remove([10,48,34])
scores.sort()
print("Sorted:", scores)
```

```python
unsorted1 = [1, 100, 24, 11]
unsorted2 = [1, 100, 24, 11]
print("Initial:")
print(unsorted1)
print(unsorted2)

print("\nsoted values:")
print(unsorted1.sort())
print(unsorted1)
print(sorted(unsorted2))

print("\nafter sort operations:")
print(unsorted1)
print(unsorted2)
```

**Key Difference: `sort()` vs `sorted()`**

* `list.sort()`

  * Modifies the original list **in place**.
  * Returns `None`.
  * The original variable is permanently changed.

* `sorted(list)`

  * Returns a **new sorted list**.
  * Does **not** modify the original list.

**In this code:**

* `unsorted1.sort()` changes `unsorted1` directly.
* `sorted(unsorted2)` prints a sorted version, but `unsorted2` remains unchanged afterward.

**After the operations:**

* `unsorted1` → sorted
* `unsorted2` → still in original order

### 2. Sets

Empty set:  

majors = set()  
⚠️ {} makes an empty dict, not a set.

```python
# Empty set
majors = set()
print("Majors:", majors)
```

```python
# Add elements
majors.add("Math")
majors.add("CS")
majors.add("Math")  # duplicate ignored
majors.add("Biology")
print("Set with duplicates added:", majors)
```

```python
# remove and discard
majors.remove("Math")
print("Set after remove:", majors)
```

```python
# remove and discard
majors.remove("Math")
print("Set after remove:", majors)
```

```python
# remove and discard
majors.discard("Math")
print("Set after remove:", majors)
```

```python
# Iterate
for major in majors:
  print(major)
```

### 3. Dictionaries

```python
# Empty dictionary
student = {}
print("Students:", student)
```

```python
# Add key/value pairs
student["name"] = "Alex"
student["gpa"] = 3.2
student["hours"] = 6
print("Populated dict:", student)
```

```python
# Update value
student["gpa"] = 3.5
print("Updated GPA:", student)
```

```python
# Access value
print("Name only:", student["name"])
```

```python
# Iterate
for key, value in student.items():
    print(key, ":", value)
```

```python
# Iterate
for key in student.keys():
    print(key)
```

```python
# Iterate
for whatever in student.values():
    print(whatever)
```
